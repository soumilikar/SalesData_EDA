# Data Cleaning

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv('shopping_trends.csv', encoding= 'unicode_escape')
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
pd.isnull(df).sum()
```

```python
df['Purchase Amount (USD)'] = df['Purchase Amount (USD)'].astype('float64')
```

```python
df['Purchase Amount (USD)'].dtypes
```

```python
df.columns
```

```python
df.rename(columns= {'Purchase Amount (USD)':'Bill (USD)'})
```

```python
df.describe()
```

```python
df[['Age', 'Previous Purchases']].describe()
```

```python
df['Age Bracket'] = '18-30'
df.loc[df['Age'] > 60, 'Age Bracket'] = '60+'
df.loc[(df['Age'] > 50) & (df['Age'] <= 60), 'Age Bracket'] = '50-60'
df.loc[(df['Age'] > 40) & (df['Age'] <= 50), 'Age Bracket'] = '40-50'
df.loc[(df['Age'] > 30) & (df['Age'] <= 40), 'Age Bracket'] = '30-40'
```

```python
df.head()
```

# Exploratory Data Analysis

## Gender

```python
ax = sns.countplot(x = 'Gender',data = df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False) ['Purchase Amount (USD)'].sum().sort_values(by = 'Purchase Amount (USD)', ascending = False)
sns.barplot(x = 'Gender', y = 'Purchase Amount (USD)', data = sales_gen,  hue = 'Gender')
```

### From this we concur, most buyers are male and their purchasing power is higher too.



## Age

```python
ax = sns.countplot(data = df, x = 'Age Bracket', hue = 'Gender')

for b in ax.containers:
    ax.bar_label(b)
```

```python
sales_age = df.groupby(['Age Bracket'], as_index=False)['Purchase Amount (USD)'].sum().sort_values(by='Purchase Amount (USD)', ascending=False)

sns.barplot(x = 'Age Bracket',y= 'Purchase Amount (USD)' ,data = sales_age)
```

### From this we concur, most of the buyers are 18-30 year old men.

## State

```python
sns.set(rc={'figure.figsize':(25,7)})
sales_state = df.groupby(['Location'], as_index=False) ['Purchase Amount (USD)'].sum().sort_values(by = 'Purchase Amount (USD)', ascending = False)
top_10 = sales_state.head(10)
bottom_10 = sales_state.tail(10)
sns.barplot(x = 'Location',y= 'Purchase Amount (USD)', data = top_10)
sns.barplot(x = 'Location',y= 'Purchase Amount (USD)', data = bottom_10)
plt.title("Top and Bottom 10 States by $", fontsize=15, fontweight="bold")
```

```python
sns.set(rc={'figure.figsize':(25,7)})
sales_state = df.groupby(['Location'], as_index=False) ['Previous Purchases'].sum().sort_values(by = 'Previous Purchases', ascending = False)
top_10 = sales_state.head(10)
bottom_10 = sales_state.tail(10)
sns.barplot(x = 'Location',y= 'Previous Purchases', data = top_10)
sns.barplot(x = 'Location',y= 'Previous Purchases', data = bottom_10)
plt.title("Top and Bottom 10 States by Orders", fontsize=15, fontweight="bold")
```

### We observe: High population states (Florida, Texas, and New Jersey) are underperforming, whereas low population states (Montana, Idaho) are leading. Hence population doesnot mean more revenue.

## Product Catagory

```python
item_data = df.groupby(['Category', 'Item Purchased']).size().reset_index(name='count').sort_values(['Category', 'count'], ascending=[True, False])
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.barplot(data=item_data, x='Item Purchased', y='count', hue='Category')

for bars in ax.containers:
    ax.bar_label(bars)
```

### We observe: Purchase count of each product, category wise, from highest to lowest. 

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Category', hue = 'Size')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Category', hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

### We can concur: Highest revenue driving catagory is Men's clothing, with size 'M' being the most common size.
